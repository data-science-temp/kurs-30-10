## Schritt 5: Datentypen korrigieren 🔢

Die Spalte `niederschlag` sollte Zahlen enthalten, wurde aber als Text eingelesen. Warum? Schauen wir nach Werten, die keine Zahlen sind:

```python
df[df["niederschlag"] == "k.A."]
```

Da steht „k.A.“ (keine Angabe). Weil in der Spalte ein Text vorkommt, behandelt Pandas die *ganze* Spalte als Text – und mit Text können wir nicht rechnen.

Mit `pd.to_numeric()` wandeln wir die Spalte in Zahlen um. Mit `errors="coerce"` sagen wir Pandas: Wenn sich ein Wert nicht umwandeln lässt, mach daraus einen fehlenden Wert (`NaN`, *not a number*).

```python
df["niederschlag"] = pd.to_numeric(df["niederschlag"], errors="coerce")
df.info()
```

## Schritt 6: Fehlwerte und Ausreißer finden 🚨

Bei `describe()` ist uns aufgefallen, dass es Temperaturen von **-999 °C** gibt. Das ist physikalisch unmöglich (der absolute Nullpunkt liegt bei -273 °C 😉). Viele Messgeräte und Datenbanken – zum Beispiel auch die des Deutschen Wetterdienstes – benutzen `-999` als Code für „Messung fehlt“.

Solche Codes müssen wir in echte fehlende Werte umwandeln. Sonst würden sie z.B. den Mittelwert völlig verfälschen!

```python
# Zeilen mit -999 anzeigen
df[(df["temp_mittel"] == -999) | (df["temp_max"] == -999)]
```

```python
import numpy as np

df["temp_mittel"] = df["temp_mittel"].replace(-999, np.nan)
```

## Schritt 7: Mit fehlenden Werten umgehen 🕳️

Wie viele Werte fehlen jetzt in jeder Spalte? `isna()` prüft für jeden Wert, ob er fehlt, und `.sum()` zählt pro Spalte:

```python
df.isna().sum()
```

Für fehlende Werte gibt es verschiedene Strategien:

| Strategie | Befehl | Wann sinnvoll? |
|---|---|---|
| **Stehen lassen** | – | Pandas ignoriert `NaN` bei `mean()`, `sum()` usw. automatisch. Oft die ehrlichste Lösung! |
| **Zeilen entfernen** | `df.dropna()` | Wenn nur wenige Zeilen betroffen sind und die Zeile ohne den Wert nutzlos ist |
| **Auffüllen** | `df.fillna(wert)` | Wenn es einen sinnvollen Ersatzwert gibt (z.B. 0, Mittelwert, Vormonat) |

Es gibt nicht *die eine* richtige Lösung. Wichtig ist, dass du bewusst entscheidest und deine Entscheidung dokumentierst.

```python
# So sieht der Datensatz aus, wenn wir alle Zeilen mit fehlenden Werten entfernen
df.dropna().shape
```
