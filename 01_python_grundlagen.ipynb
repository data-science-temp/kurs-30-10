## Datentypen 🔢

Python unterscheidet verschiedene Arten von Werten, sogenannte **Datentypen**. Die wichtigsten für uns sind:

| Datentyp | Bedeutung | Beispiel |
|---|---|---|
| `int` | ganze Zahl (*integer*) | `2024` |
| `float` | Kommazahl | `9.4` |
| `str` | Text (*string*) | `"München"` |
| `bool` | Wahrheitswert | `True` oder `False` |

‼️ **Achtung:** In Python schreibt man Kommazahlen mit einem **Punkt**, nicht mit einem Komma: `9.4` statt `9,4`.

Mit `type()` kannst du herausfinden, welchen Datentyp ein Wert hat:

````python
print(type(2024))
print(type(9.4))
print(type("München"))
print(type(True))
````

Mit Zahlen kann man ganz normal rechnen: `+`, `-`, `*` (mal), `/` (geteilt) und `**` (hoch). Mit `round()` kannst du ein Ergebnis runden.

````python
jahre = 2024 - 1990
print(jahre)

print(10 / 3)
print(round(10 / 3, 2))  # auf 2 Nachkommastellen runden
````

## Listen 📋

Eine **Liste** speichert mehrere Werte in einer einzigen Variable. Die Werte stehen in eckigen Klammern und sind durch Kommas getrennt. Hier sind die Höchsttemperaturen einer (erfundenen) Sommerwoche:

````python
temperaturen = [24.1, 27.5, 31.2, 29.8, 22.4, 25.0, 30.6]
print(temperaturen)
````

Bei der Abfrage einzelner Elemente müssen wir aufpassen: Der Computer fängt **bei 0 an zu zählen**! Mit `-1` bekommst du das letzte Element.

Außerdem gibt es ein paar praktische Funktionen: `len()` gibt die Länge der Liste aus, `sum()` die Summe, `min()` und `max()` den kleinsten und größten Wert.

````python
print(temperaturen[0])   # 1. Element
print(temperaturen[2])   # 3. Element
print(temperaturen[-1])  # letztes Element

print(len(temperaturen))
print(max(temperaturen))
````

Mit `.append()` kannst du ein neues Element ans Ende der Liste anhängen:

````python
temperaturen.append(26.3)
print(temperaturen)
````

## Dictionaries 📖

Ein **Dictionary** (Wörterbuch) speichert Paare aus einem *Schlüssel* und einem *Wert*. So kann man zusammengehörige Informationen übersichtlich ablegen. Dictionaries stehen in geschweiften Klammern `{ }`:

````python
stadt_info = {
    "name": "München",
    "bundesland": "Bayern",
    "einwohner": 1510000,
    "hoehe_m": 519
}

print(stadt_info["name"])
print(stadt_info["einwohner"])
````

Einen neuen Eintrag fügt man so hinzu: `stadt_info["fluss"] = "Isar"`.

## Bedingungen 🔀

Oft wollen wir, dass unser Programm Entscheidungen trifft. Dafür gibt es `if`, `elif` und `else`:

```
if (Bedingung ist erfüllt):
    dieser Code wird ausgeführt
elif (andere Bedingung ist erfüllt):
    dieser Code wird ausgeführt
else:
    sonst wird dieser Code ausgeführt
```

Wichtig sind der **Doppelpunkt** und die **Einrückung** danach! Zum Vergleichen gibt es `<`, `<=`, `>`, `>=`, `==` (gleich) und `!=` (ungleich). Mehrere Bedingungen kann man mit `and` und `or` verbinden.

Der Deutsche Wetterdienst nennt einen Tag mit mindestens 25 °C einen *Sommertag* und einen Tag mit mindestens 30 °C einen *heißen Tag*:

````python
temperatur = 27.5

if temperatur >= 30:
    print("Heißer Tag! 🥵")
elif temperatur >= 25:
    print("Sommertag ☀️")
else:
    print("Normaler Tag")
````

## Schleifen 🔁

Eine **Schleife** wiederholt einen Teil deines Programms. Mit einer `for`-Schleife kannst du zum Beispiel jedes Element einer Liste nacheinander durchgehen:

````python
for t in temperaturen:
    print(t, "Grad")
````

In Worten sagt dieses Programm: *Für jeden Wert `t` in der Liste `temperaturen`, gib `t` und das Wort „Grad“ aus.*

Mit `range()` kannst du dir Zahlenfolgen erzeugen: `range(5)` ergibt 0, 1, 2, 3, 4 und `range(1990, 1995)` ergibt 1990 bis 1994.

````python
for jahr in range(1990, 1995):
    print(jahr)
````

Schleifen und Bedingungen lassen sich gut kombinieren. Hier zählen wir mit einer Zähler-Variable, wie viele Tage wärmer als 28 Grad waren:

````python
anzahl = 0
for t in temperaturen:
    if t > 28:
        anzahl = anzahl + 1

print("Tage über 28 Grad:", anzahl)
````

Neben der `for`-Schleife gibt es auch die `while`-Schleife. Sie läuft so lange weiter, wie eine Bedingung erfüllt ist. Achtung: Wenn die Bedingung nie falsch wird, haben wir eine Endlos-Schleife!

````python
jahr = 2020
while jahr < 2025:
    print(jahr)
    jahr = jahr + 1
````

## Funktionen 🧩

Falls ihr es noch nicht gemerkt habt: Programmierer:innen sind recht faul und wollen so wenig doppelte Arbeit wie möglich 😉. Mit **Funktionen** kannst du Code einmal schreiben und dann immer wieder verwenden.

Man definiert eine Funktion mit dem Schlüsselwort `def`. Werte, die man der Funktion mitgibt, heißen *Parameter*. Mit `return` gibt die Funktion ein Ergebnis zurück:

````python
def celsius_zu_fahrenheit(celsius):
    return celsius * 1.8 + 32

# Funktionsaufruf
print(celsius_zu_fahrenheit(9.4))
print(celsius_zu_fahrenheit(30))
````

Übrigens: `print`, `len`, `sum` und `round` sind auch Funktionen – nur dass sie schon in Python eingebaut sind!
