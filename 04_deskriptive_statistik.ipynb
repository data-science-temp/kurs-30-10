## Lage: Wo liegt die Mitte? 🎯

Die zwei wichtigsten Kennzahlen für die „Mitte“ von Daten sind:

* **Mittelwert** (*mean*): Summe aller Werte geteilt durch die Anzahl. Das kennst du als „Durchschnitt“.
* **Median**: Der Wert genau in der Mitte, wenn man alle Werte der Größe nach sortiert. Die Hälfte der Werte ist kleiner, die andere Hälfte größer.

Der Unterschied wird an einem kleinen Beispiel klar. Fünf Studentinnen geben an, wie viele Minuten sie zur Uni pendeln:

```python
pendelzeit = pd.Series([15, 20, 25, 30, 180])

print("Mittelwert:", pendelzeit.mean())
print("Median:", pendelzeit.median())
```

Eine Person pendelt drei Stunden, und schon steigt der Mittelwert auf 54 Minuten, obwohl vier von fünf Personen höchstens 30 Minuten brauchen. Der Median (25) bleibt davon unbeeindruckt.

💡 **Merke:** Der Mittelwert ist empfindlich gegenüber Ausreißern, der Median ist *robust*. Wenn Mittelwert und Median weit auseinanderliegen, lohnt es sich, genauer hinzuschauen!

In Pandas funktioniert das genauso für jede Spalte:

```python
print(df["temp_mittel"].mean())
print(df["temp_mittel"].median())
```

## Streuung: Wie stark schwanken die Werte? 📏

Zwei Städte können die gleiche Durchschnittstemperatur haben, aber trotzdem ganz unterschiedlich sein: In der einen ist es das ganze Jahr mild, in der anderen gibt es heiße Sommer und eiskalte Winter. Deshalb schauen wir uns auch die **Streuung** an:

* **Minimum** und **Maximum**: der kleinste und der größte Wert. Ihre Differenz heißt **Spannweite**.
* **Standardabweichung** (*standard deviation*, `std`): Wie weit liegen die Werte im Schnitt vom Mittelwert entfernt? Eine kleine Standardabweichung heißt: Die Werte liegen eng beieinander.
* **Quartile**: Teilen die sortierten Daten in vier gleich große Teile. 25 % der Werte liegen unter dem 1. Quartil, 75 % unter dem 3. Quartil.

```python
print("Minimum:", df["temp_mittel"].min())
print("Maximum:", df["temp_mittel"].max())
print("Standardabweichung:", df["temp_mittel"].std())
print("1. Quartil:", df["temp_mittel"].quantile(0.25))
print("3. Quartil:", df["temp_mittel"].quantile(0.75))
```

All diese Werte auf einmal liefert dir übrigens `describe()`, das wir schon kennen. Jetzt weißt du, was die Zeilen bedeuten! `50%` ist dabei der Median.

```python
df["temp_mittel"].describe()
```

## Gruppieren mit `groupby()` 🧺

Bisher haben wir für jede Stadt einzeln gefiltert. Das wird bei vielen Städten schnell mühsam. Mit `groupby()` geht es auf einmal: Pandas teilt die Daten in Gruppen auf, berechnet die Kennzahl für jede Gruppe und fügt das Ergebnis wieder zusammen.

Das Muster lautet: `df.groupby("Gruppenspalte")["Wertespalte"].Kennzahl()`

```python
df.groupby("stadt")["temp_mittel"].mean()
```

Mit `.round()` wird es übersichtlicher, und mit `sort_values()` können wir das Ergebnis direkt sortieren:

```python
df.groupby("stadt")["temp_mittel"].mean().round(1).sort_values(ascending=False)
```
