## Spalten auswählen 🎯

Eine einzelne Spalte wählst du mit eckigen Klammern und dem Spaltennamen aus – ganz ähnlich wie bei einem Dictionary:

```python
df[["stadt", "monat", "niederschlag"]]
```

Auf einzelnen Spalten kannst du direkt rechnen, z.B. mit `.max()`, `.min()` oder `.sum()`. Mit `.unique()` siehst du alle unterschiedlichen Werte einer Spalte:

## Filtern 🔎

Oft interessieren uns nur bestimmte Zeilen, z.B. nur die Werte für München. Dafür schreiben wir eine **Bedingung** in die eckigen Klammern – genau wie bei `if` mit `==`, `>`, `<` usw.:

```python
df[df["stadt"] == "München"]
```

Was passiert hier? `df["stadt"] == "München"` prüft für jede Zeile, ob die Bedingung stimmt, und ergibt `True` oder `False`. Pandas behält dann nur die Zeilen mit `True`.

Mehrere Bedingungen verbindest du mit `&` (und) bzw. `|` (oder). Jede Bedingung muss dann in runden Klammern stehen:

```python
# Alle Monate in München mit mehr als 100 mm Niederschlag
df[(df["stadt"] == "München") & (df["niederschlag"] > 100)]
```
